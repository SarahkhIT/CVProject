### Training Interpretation

We fine-tuned `yolov8n-seg.pt` on our Roboflow "Solid Waste" dataset (9 classes,
113 training images, 10 validation images) for up to 60 epochs, with the first
10 backbone layers frozen. Training stopped early via EarlyStopping at epoch 54
(patience=15), with the best checkpoint saved at **epoch 39** — box_loss and
seg_loss both declined steadily through training (box_loss: 1.35 → ~1.0, seg_loss:
3.40 → ~2.2) but plateaued in the final ~15 epochs without further validation
improvement, triggering the early stop.

The final validation results (`best.pt`) were **mAP50 = 0.319 / mAP50-95 = 0.228**
(box) and **mAP50 = 0.309 / mAP50-95 = 0.201** (mask). Rather than classic
overfitting (train loss dropping while val loss rises), our results point to
**underfitting driven by severe class imbalance**: per-class performance varied
enormously — **Glass** (45 of 64 validation instances) reached a strong mAP50 of
0.69, and **Plastic** (4 instances) reached 0.55, but **General trash** and
**Plastic bag** (2–5 instances each) scored near zero (mAP50 ≈ 0.02). With only
113 training images spread across 9 classes, minority classes simply didn't have
enough examples for the model to learn reliable features, while the dominant
class (Glass) learned well.

**Takeaway:** the pipeline and training process work correctly, but the dataset
is too small and imbalanced to support all 9 classes. A stronger result would
need either significantly more images per minority class, class-balanced
sampling/augmentation, or consolidating rare classes into a single generalized
category.

# **Real-World Solution & Video Analytics**

```python
import os
import urllib.request
import cv2
import yt_dlp
from ultralytics import YOLO
from ultralytics.solutions import ObjectCounter

VIDEO_URL = "https://youtu.be/WgJd7u8a00Y"
MODEL_URL = "https://github.com/SarahkhIT/CVProject/raw/refs/heads/main/best.pt"
LOCAL_VIDEO_PATH = "input_video.mp4"
MODEL_PATH = "best.pt"
OUTPUT_PATH = "output_sorted.mp4"
COUNTING_REGION = [(0, 360), (1280, 360)]
SHOW_VIDEO = False


def download_video(url, output_path):
    if os.path.exists(output_path):
        return output_path

    ydl_opts = {
        "format": "best[ext=mp4]/best",
        "outtmpl": output_path,
        "extractor_args": {
            "youtube": {
                "player_client": ["android"],
            }
        },
    }
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        ydl.download([url])

    return output_path


def download_model(url, output_path):
    if os.path.exists(output_path):
        return output_path

    urllib.request.urlretrieve(url, output_path)
    return output_path


def load_model(path):
    return YOLO(path)


def open_video(path):
    cap = cv2.VideoCapture(path)
    assert cap.isOpened(), f"Could not open video: {path}"
    return cap


def get_video_properties(cap):
    w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS) or 30
    return w, h, fps


def create_writer(path, fps, size):
    return cv2.VideoWriter(path, cv2.VideoWriter_fourcc(*"mp4v"), fps, size)


def create_counter(model_path, region):
    return ObjectCounter(show=False, region=region, model=model_path)


def process_video(cap, counter, writer, show):
    frame_count = 0
    last_results = None

    while cap.isOpened():
        success, frame = cap.read()
        if not success:
            break

        frame_count += 1
        results = counter(frame)
        annotated_frame = results.plot_im
        writer.write(annotated_frame)
        last_results = results

        if show:
            cv2.imshow("Smart Waste Sorting - Live Analytics", annotated_frame)
            if cv2.waitKey(1) & 0xFF == ord("q"):
                break

    return frame_count, last_results


def cleanup(cap, writer):
    cap.release()
    writer.release()
    cv2.destroyAllWindows()


def print_summary(frame_count, last_results, output_path):
    print(f"\nProcessing complete. {frame_count} frames processed.")
    if last_results is not None:
        print(f"Total objects counted (in): {getattr(last_results, 'in_count', 'N/A')}")
        print(f"Total objects counted (out): {getattr(last_results, 'out_count', 'N/A')}")
        print(f"Per-class counts: {getattr(last_results, 'classwise_count', 'N/A')}")
    else:
        print("No frames were processed — no results to summarize.")
    print(f"Output video saved to: {output_path}")


def main():
    video_path = download_video(VIDEO_URL, LOCAL_VIDEO_PATH)
    model_path = download_model(MODEL_URL, MODEL_PATH)

    model = load_model(model_path)
    cap = open_video(video_path)
    w, h, fps = get_video_properties(cap)

    print(f"Video resolution: {w}x{h}")

    region = [(0, h // 2), (w, h // 2)]

    writer = create_writer(OUTPUT_PATH, fps, (w, h))
    counter = create_counter(model_path, region)

    frame_count, last_results = process_video(cap, counter, writer, SHOW_VIDEO)

    cleanup(cap, writer)
    print_summary(frame_count, last_results, OUTPUT_PATH)


if __name__ == "__main__":
    main()
```

```python
from ultralytics import YOLO
from PIL import Image
import os

eval_model = YOLO("best.pt")

metrics = eval_model.val(
    data=f"{dataset.location}/data.yaml",
    imgsz=640,
    conf=0.25,
    iou=0.70,
    plots=True
)

print("Bounding Box Metrics")
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)

print("\nSegmentation Metrics")
print("Precision:", metrics.seg.mp)
print("Recall:", metrics.seg.mr)
print("mAP50:", metrics.seg.map50)
print("mAP50-95:", metrics.seg.map)

confusion_path = os.path.join(
    str(metrics.save_dir),
    "confusion_matrix_normalized.png"
)

if os.path.exists(confusion_path):
    display(Image.open(confusion_path))

confidence_values = [0.15, 0.25, 0.40, 0.50]

for conf_value in confidence_values:
    result = eval_model.val(
        data=f"{dataset.location}/data.yaml",
        imgsz=640,
        conf=conf_value,
        iou=0.70,
        plots=False,
        verbose=False
    )

    print("\nConfidence:", conf_value)
    print("Precision:", result.seg.mp)
    print("Recall:", result.seg.mr)
    print("mAP50:", result.seg.map50)
```

## Evaluation Analysis

The model was evaluated using the validation dataset with Precision, Recall, mAP50, and mAP50-95.

The confusion matrix was used to identify prediction errors and class confusion. False positives represent incorrect waste predictions, while false negatives represent waste objects that the model failed to detect.

Different confidence thresholds were tested to compare the trade-off between Precision and Recall. Lower confidence thresholds may increase Recall but can also increase false positives, while higher confidence thresholds may improve Precision but miss more objects.

A confidence threshold was selected based on the best balance between Precision and Recall for the waste sorting task.

An IoU threshold of 0.70 was used to evaluate how well the predicted bounding boxes and segmentation masks overlap with the ground-truth annotations.

### Why ONNX?
We chose **ONNX (Open Neural Network Exchange)** as our export format because it
provides cross-platform compatibility, allowing the model to run on different
hardware and inference runtimes (CPU, GPU, edge devices) without depending on the
original PyTorch environment. ONNX also benefits from graph-level optimizations
that can improve inference speed, which is valuable for a waste-sorting system
that may eventually need to run in real time on lower-power hardware.
